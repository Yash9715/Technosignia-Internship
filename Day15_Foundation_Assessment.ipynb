## 1. Import libraries and load data
Keep the CSV file in the same folder as this notebook.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import sqlite3

df = pd.read_csv("sample_sales_dataset.csv")
df.head()
```

## 2. Python and Pandas

```python
print("Rows:", len(df))
print("Columns:", len(df.columns))
print("\nData types:")
print(df.dtypes)
print("\nMissing values:")
print(df.isnull().sum())
print("\nDuplicate rows:", df.duplicated().sum())
```

```python
df["Profit_Margin_Percent"] = np.where(
    df["Sales"] != 0, df["Profit"] / df["Sales"] * 100, 0
)
high_value_orders = df[df["Sales"] >= 10000]
print("High-value orders:", len(high_value_orders))
display(high_value_orders[["Order_ID","Product","Sales","Profit"]].head(10))
```

## 3. Statistics

```python
display(df[["Quantity","Unit_Price","Sales","Profit"]].describe())
print("Mean sales:", round(df["Sales"].mean(), 2))
print("Median sales:", round(df["Sales"].median(), 2))
print("Sales standard deviation:", round(df["Sales"].std(), 2))
print("Minimum sales:", df["Sales"].min())
print("Maximum sales:", df["Sales"].max())
```

## 4. SQL assessment using SQLite

```python
conn = sqlite3.connect(":memory:")
df.to_sql("sales", conn, index=False, if_exists="replace")

query = '''
SELECT Category,
       SUM(Sales) AS total_sales,
       SUM(Profit) AS total_profit
FROM sales
GROUP BY Category
ORDER BY total_sales DESC;
'''
category_result = pd.read_sql_query(query, conn)
display(category_result)
```

### SQL query explanation for mentor
- `SELECT` chooses the output columns.
- `SUM(Sales)` and `SUM(Profit)` calculate category totals.
- `FROM sales` reads the sales table.
- `GROUP BY Category` groups records by product category.
- `ORDER BY total_sales DESC` sorts the categories from highest to lowest sales.

**Business use:** Identify which product category contributes the most revenue.


## 5. Visualization

```python
category_sales = df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
plt.figure(figsize=(8,5))
category_sales.plot(kind="bar")
plt.title("Total Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

### Chart explanation for mentor
The bar chart compares total sales across product categories. The tallest bar represents the category generating the highest revenue. This comparison can support inventory planning and marketing decisions. Refer to the chart output after running the notebook.


## 6. Business observations

```python
print("Total sales:", df["Sales"].sum())
print("Total profit:", df["Profit"].sum())
print("Highest revenue category:", category_sales.idxmax())
print("Highest revenue city:", df.groupby("City")["Sales"].sum().idxmax())
print("Online sales:", df.loc[df["Channel"]=="Online","Sales"].sum())
print("Offline sales:", df.loc[df["Channel"]=="Offline","Sales"].sum())
```

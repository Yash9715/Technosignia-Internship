### file creating and writting

```python
with open ('new_file.txt','w',) as file:
    file.write ("hello! User\n")
    file.write("you can continue your writing")
print('file writting completed.')    
    
```

### file reading

```python
with open ('new_file.txt','r')as file:
    content=file.read()
print(content)    
```

```python
with open ('new_file.txt','r')as file:
    content=file.readline()
    content2=file.readlines()
print(content)    
print(content2)    
```

### appending to the file

```python
with open ('new_file.txt','a') as file:
    file.append("india is my country")
print('content added sucessfully!')    
```

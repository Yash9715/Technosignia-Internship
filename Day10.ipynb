### Statistics_Data_Analytics

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
print("Libraries imported successfully")
```

```python
marks =[10,20,20,30,40,50,60,70,80,1000]

df=pd.DataFrame({
    "Student_ID":range(1,11),
    "Marks":marks
})
display(df)
```

### Mean — Average

```python
mean_value = np.mean(marks)
print("Mean",mean_value)
```

### Median
### Median is the middle value of sorted data. For an even number of observations, it is the average of the two middle values.

```python
median_value = np.median(marks)
print("Median",median_value)
```

### Mode
### Most repeated data 

```python
mode_value= df["Marks"].mode()
print("Mode",mode_value.tolist())
```

### Variance 
### Variance measures the average squared distance of every data point from the mean.

### STD.Deviation
### Because variance uses squared units, it can be hard to interpret intuitively. Standard deviation solves this by taking the square root of the variance.

```python
population_variance=np.var(marks)
population_std=np.std(marks)

print("Variance",round(population_variance,2))
print("std",round(population_std,2))
```

### Quartiles
### Quartiles split your sorted dataset into four equal parts (25% chunks).

```python
q1=np.percentile(marks,25)
q2=np.percentile(marks,50)
q3=np.percentile(marks,75)

p90=np.percentile(marks,90)

print("Q1 (25th percentile):", q1)
print("Q2 (50th percentile):", q2)
print("Q3 (75th percentile):", q3)
print("90th percentile:", p90)
```

### Outlier 
### An outlier is a data point that lies an abnormal distance from other values in a random sample from a population.

```python
q1=np.percentile(marks,25)
q3=np.percentile(marks,75)

iqr=q3-q1
lower_bound=q1-1.5*iqr
upper_bound=q3+1.5*iqr

outlier=[
    value for value in marks
    if value < lower_bound or value > upper_bound
]
print("IQR",iqr)
print("Lower_bound:",lower_bound)
print("Upper_bound:",upper_bound)
print("Outliers",outlier)
```

### Distribution and Charts

### Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(marks,bins=8,edgecolor='black')

plt.title("Distribution of students marks")
plt.xlabel('marks')
plt.ylabel('frequency')
plt.show()
```

### Boxplot

```python
plt.figure(figsize=(8,4))
plt.boxplot(marks,orientation="horizontal")

plt.title("Boxplot of students marks")
plt.xlabel('marks')
plt.show()
```

### Correlation

```python
study_data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6, 7, 8],
    "Marks": [35, 40, 50, 55, 65, 70, 80, 90]
})

correlation = study_data["Study_Hours"].corr(
    study_data["Marks"]
)

print("Correlation:", round(correlation, 3))
```

### Scatterplot

```python
plt.figure(figsize=(7, 5))

plt.scatter(
    study_data["Study_Hours"],
    study_data["Marks"]
)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```

```python
original_marks = [10, 20, 20, 30, 40, 50, 60, 70, 80, 1000]

cleaned_marks = [
    value for value in original_marks
    if value != 1000
]

comparison = pd.DataFrame({
    "Measure": ["Mean", "Median"],
    "With Outlier": [
        np.mean(original_marks),
        np.median(original_marks)
    ],
    "Without Outlier": [
        np.mean(cleaned_marks),
        np.median(cleaned_marks)
    ]
})

display(comparison.round(2))
```

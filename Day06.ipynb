# Day 6 — Pandas I: Data Loading and Exploration
**Dataset:** `Day13_Sales_100_Records.csv` (keep in same folder as notebook)  
**Deliverable:** An exploratory notebook and a data dictionary describing each column.

```python
import pandas as pd

df = pd.read_csv("Day13_Sales_100_Records.csv")
print("Pandas version:", pd.__version__)
print("Rows, columns:", df.shape)
display(df.head())
```

## 1. Load and inspect data

```python
display(df.head())
display(df.tail())
df.info()
display(df.describe())
```

## 2. Rows, columns, types, and selecting columns

```python
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Column names:", df.columns.tolist())
display(df.dtypes.to_frame("Data Type"))

display(df[["Product", "Quantity", "Sales"]].head())
display(df.loc[0:4, ["OrderID", "OrderDate", "Region", "Sales"]])
```

## 3. Missing values and possible quality issues

```python
display(df.isna().sum().to_frame("Missing Values"))
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate OrderIDs:", df["OrderID"].duplicated().sum())
print("Date strings sample:", df["OrderDate"].head().tolist())
```

## 4. Data dictionary

```python
data_dictionary = pd.DataFrame({
    "Column": df.columns,
    "Meaning": [
        "Unique identifier for a sales order",
        "Date the order was placed",
        "Product name",
        "Broad product category",
        "Sales region",
        "Sales channel",
        "Customer segment",
        "Number of units sold",
        "Price per unit before discount",
        "Discount rate as a decimal",
        "Sales value after discount"
    ],
    "Expected Type": [
        "Text", "Date", "Category/Text", "Category/Text", "Category/Text",
        "Category/Text", "Category/Text", "Integer", "Numeric", "Decimal",
        "Numeric"
    ]
})
display(data_dictionary)
```

### Arithmatic Operator

```python
a=30
b=7
print(a+b) #addition
print(a-b) #subtraction
print(a*b) #multiplication
print(a/b) #division
print(a//b) #floor division
```

### Logical Operator

```python
print(a>b)  #greater than check
print(a>=b)  #greater than equal to check
print(a<b) #smaller than check
print(b<a) #smaller than equal to check
print(a==b) #is equal to check
print(a!=b) #not equal to check

```

### Logical Operator

```python
name='Ram'
salary=80000
adress='Pune'
print(name=='Ram' and adress=='Pune')
print(salary == 80000 and adress=='Mumbai')
print(name=='Ram' or adress=='Mumbai')
print(salary == 80000)
```

### Programs for calculating percentages, sales totals, discounts, and averages.  


```python
math=int(input("Enter your maths marks "))
marathi=int(input("Enter your marathi marks "))
science=int(input("Enter your science marks "))
history=int(input("Enter your history marks "))
english=int(input("Enter your english marks "))
hindi=int(input("Enter your hindi marks "))
total_marks=math+marathi+science+history+english+hindi
percentage=(total_marks/600)*100
print(f"Your Percentage is {percentage:.2f}")
```

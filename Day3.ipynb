### Conditional Statements (IF,ELIF,IF-ELSE)

```python
marks=int(input('Enter your marks:'))
if marks>= 45:                         #if condition
    print("you passed ")

age=int(input("Enter your age "))
if age>=18:
    print("you are eligible for voting")     #if-else condition
else:
    print("you are not eligible")


per=int(input('Enter your current percentage'))
if per>=90:
    print("grade=A")
elif per>=80:
    print("you selected for scolarship")
    print('grade=B')
elif per>=70:
    print("you selected for scolarship")   #if-elif-else condition 
    print('grade=C')    
elif per>=50:
    print("you selected for scolarship")
    print('grade=D')
else:
    print("you failed for scolarship")
```

### Loop (FOR,WHILE LOOP)

```python
for n in range(6):
    print(n)
print()    

n=int(input("Enter the number whose table you want"))
for i in range (1,11):
    print(f"{n} * {i} = {n*i}")                          #for loop
print()

lst=['banana','grapes','apples','guava','mangoes']
for fruit in lst:
    print(fruit)
```

### while loop

```python
n=5
while n>0:
    print(n)
    n=n-1
print()
                              #while loop
s=0
n=int(input("Enter the number this will give you addition upto that number:"))
while(n>0):
    s+=n
    n-=1
print(s)
```

### Python Collections (LIST,TUPLE,SET,DICTONARY)

```python
list=['ram',101,'sita',4.5,True,55,345,43.3]
print(list)
print(list[0],list[1])              #list
list[4]=False
print(list)
list.append(55)
print(list)
list2=['rohit','aaja',343]
list.extend(list2)
print(list)
list.remove('aaja')
print(list)

```

```python
marks=(66,76,54,55,33,65,76,33)
print('total subject are :',len(marks))
print('sum of marks is :',sum(marks))
print('highest marks is :',max(marks))             #Tuple
print('lowest marks is :',min(marks))
print('index of 33 marks is :',marks.index(33))
```

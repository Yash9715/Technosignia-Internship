# Day 5 — NumPy for Numerical Computing
**Objective:** Practice arrays, indexing, slicing, filtering, reshaping, and numerical statistics.

**Deliverable:** A notebook demonstrating NumPy array operations and numerical calculations.

```python
import numpy as np

print("NumPy version:", np.__version__)
```

## 1. Create one-dimensional and two-dimensional arrays

```python
sales = np.array([1200, 2500, 1800, 3200, 2800, 4100])
print("Sales:", sales)
print("Shape:", sales.shape)
print("Dimensions:", sales.ndim)
print("Data type:", sales.dtype)

matrix = np.array([[10, 20, 30], [40, 50, 60]])
print("\n2D array:\n", matrix)
print("2D shape:", matrix.shape)
```

## 2. Indexing and slicing

```python
print("First value:", sales[0])
print("Last value:", sales[-1])
print("First three values:", sales[:3])
print("Every second value:", sales[::2])
print("Second row, third column:", matrix[1, 2])
```

## 3. Boolean filtering

```python
high_sales = sales[sales > 2500]
print("Sales above 2500:", high_sales)
```

## 4. Reshape arrays

```python
values = np.arange(1, 13)
grid = values.reshape(3, 4)
print(grid)
print("New shape:", grid.shape)
```

## 5. Numerical calculations

```python
print("Sum:", sales.sum())
print("Mean:", sales.mean())
print("Minimum:", sales.min())
print("Maximum:", sales.max())
print("Standard deviation:", sales.std())
```

## 6. Compare Python and NumPy calculations

```python
python_values = [1200, 2500, 1800, 3200, 2800, 4100]
python_total = sum(python_values)
numpy_total = np.array(python_values).sum()

print("Python total:", python_total)
print("NumPy total:", numpy_total)
print("Same result:", python_total == numpy_total)
```
